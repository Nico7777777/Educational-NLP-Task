```python
import pandas as pd
```

```python
disc = pd.read_csv("datasets/discussion.csv")
prob = pd.read_csv("datasets/problemset.csv")
print(disc.shape)
```

```python
first_tier = disc.loc[disc['Parent_Id']==0, :]
second_tier = disc.loc[disc['Parent_Id']!=0, :]
print(first_tier.shape)
print(second_tier.shape)
```

```python
first_tier_sample = first_tier.sample(500)
second_tier_sample = second_tier.sample(500)
```

### Creez primul sample de adnotari

```python
# sentences = pd.DataFrame([prob.loc[prob['Problem Name']==text, 'Description'] for text in first_tier_sample['Problem Name']])
first_tier_sample = first_tier_sample.merge(prob[['Problem Name', 'Description']], on='Problem Name', how='left')
```

```python
print(first_tier_sample.head(n=4))
first_tier_sample.to_csv('annotations/first_tier_sample.csv')
```

### Creez al doilea set de adnotari

```python
print(second_tier_sample.index)  # Verifică indexul lui second_tier_sample
print(disc.index)  # Verifică indexul lui disc
second_tier_sample.reset_index(drop=True, inplace=True)
disc.reset_index(drop=True, inplace=True)
```

```python
second_tier_sample['Parent Text'] = disc.loc[
    (disc['Id'] == second_tier_sample['Parent_Id']) &
    (disc['Problem Name'] == second_tier_sample['Problem Name']),
    'Text'
]
second_tier_sample.head(n=4)
# merged = second_tier.merge(disc, left_on=['Parent_Id', 'Problem Name'], right_on=['Id', 'Problem Name'], how='left')
# merged.head(n=10)
```

```python
second_tier_sample = second_tier_sample.merge(
    disc[['Id', 'Problem Name', 'Text']],
    left_on=['Parent_Id', 'Problem Name'],
    right_on=['Id', 'Problem Name'],
    how='inner'
)
# second_tier_sample.head(n=10)
```

```python
second_tier_sample.drop(columns=['Parent_Text'], inplace=True)
second_tier_sample.rename(columns={'Text_y':'Parent Text'}, inplace=True)
second_tier_sample.drop(columns=['Id_y'], inplace=True)
```

```python
merged = second_tier_sample.merge(
    disc[['Id', 'Problem Name', 'Up Votes']],
    left_on=['Parent_Id', 'Problem Name'],
    right_on=['Id', 'Problem Name'],
    how='left'
)
second_tier_sample['Parent_Up_Votes'] = merged['Up Votes_y']
```

```python
merged = second_tier_sample.merge(
    disc[['Id', 'Problem Name', 'Date']],
    left_on=['Parent_Id', 'Problem Name'],
    right_on=['Id', 'Problem Name'],
    how='left'
)
second_tier_sample['Parent_Date'] = merged['Date_y']
```
